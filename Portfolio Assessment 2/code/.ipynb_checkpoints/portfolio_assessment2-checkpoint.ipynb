# Step 2: Composite Columns

```python
df = pd.read_csv('combined_data_step1.csv')
```

```python
df.info()
```

```python
df.shape
```

```python
df = df.rename(columns={
    'Frame': 'frame',
    'Right Upper Leg x': 'right_leg_x',
    'Right Upper Leg y': 'right_leg_y',
    'Right Upper Leg z': 'right_leg_z',
    'Left Upper Leg x': 'left_leg_x',
    'Left Upper Leg y': 'left_leg_y',
    'Left Upper Leg z': 'left_leg_z'
})

# 1. Root mean square value of x and y
df['rms_xy_right'] = np.sqrt((df['right_leg_x']**2 + df['right_leg_y']**2) /2)
df['rms_xy_left'] = np.sqrt((df['left_leg_x']**2 + df['left_leg_y']**2) /2)

# 2. Root mean square value of y and z
df['rms_yz_right'] = np.sqrt((df['right_leg_y']**2 + df['right_leg_z']**2) /2)
df['rms_yz_left'] = np.sqrt((df['left_leg_y']**2 + df['left_leg_z']**2) /2)

# 3. Root mean square value of z and x
df['rms_zx_right'] = np.sqrt((df['right_leg_z']**2 + df['right_leg_x']**2) /2)
df['rms_zx_left'] = np.sqrt((df['left_leg_z']**2 + df['left_leg_x']**2) /2)

# 4. Root mean square value of x y z
df['rms_xyz_right'] = np.sqrt((df['right_leg_x']**2 + df['right_leg_y']**2 + df['right_leg_z']**2) / 3)
df['rms_xyz_left'] = np.sqrt((df['left_leg_x']**2 + df['left_leg_y']**2 + df['left_leg_z']**2) / 3)

# 5. Roll calculation
df['roll_right'] = 180 * np.arctan2(df['right_leg_y'], 
                                     np.sqrt(df['right_leg_x']**2 + df['right_leg_z']**2)) / np.pi
df['roll_left'] = 180 * np.arctan2(df['left_leg_y'], 
                                    np.sqrt(df['left_leg_x']**2 + df['left_leg_z']**2)) / np.pi
# 6. Pitch calculation
df['pitch_right'] = 180 * np.arctan2(df['right_leg_x'], 
                                      np.sqrt(df['right_leg_y']**2 + df['right_leg_z']**2)) / np.pi
df['pitch_left'] = 180 * np.arctan2(df['left_leg_x'], 
                                     np.sqrt(df['left_leg_y']**2 + df['left_leg_z']**2)) / np.pi

df = df.to_csv('combined_data_step2.csv',index=False)
```

# Step 3: Data Preprocessing & Feature Computation

```python
df = pd.read_csv('combined_data_step2.csv')
```

```python
df.info()
```

```python
df.shape
```

```python
def compute_features(df_segment, column_name):
    features = {}
    
    # Mean
    features[f'mean_{column_name}'] = df_segment[column_name].mean()
    
    # Standard deviation
    features[f'std_{column_name}'] = df_segment[column_name].std()
    
    # Min and Max
    features[f'min_{column_name}'] = df_segment[column_name].min()
    features[f'max_{column_name}'] = df_segment[column_name].max()
    
    # Area under the curve (AUC) - using trapezoidal rule
    features[f'auc_{column_name}'] = integrate.trapz(df_segment[column_name].values)

    # Number of peaks
    # Using prominence parameter to avoid detecting noise as peaks
    peaks, _ = find_peaks(df_segment[column_name].values, prominence=0.1)
    features[f'peaks_{column_name}'] = len(peaks)
    
    return features
    
# Identify data columns to compute features 
feature_columns = [col for col in df.columns if col not in ['frame', 'class']]

feature_rows = []

# Process data in one-minute chunks (60 frames per minute)
for class_value in [0, 1]: 
    class_df = df[df['class'] == class_value]
    
    for i in range(0, len(class_df), 60):
        minute_df = class_df.iloc[i:i+60]
        if len(minute_df) < 60:  
            continue

        # Initialize a feature dictionary for this minute
        minute_features = {'class': class_value}
        
        # Calculate features for each column
        for column in feature_columns:
            column_features = compute_features(minute_df, column)
            minute_features.update(column_features)
        
        feature_rows.append(minute_features)

feature_df = pd.DataFrame(feature_rows)

feature_df.to_csv('combined_data_step3.csv', index=False)
```

# Step 4: Training 

```python
features_df = pd.read_csv('combined_data_step3.csv')
```

```python
features_df.info()
```

```python
features_df.shape
```

```python
X = features_df.drop('class', axis=1)
y = features_df['class']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

def evaluate_model(model, X_test, y_test, model_name):
    y_pred = model.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    print(f"\n{model_name} Results:")
    print(f"Accuracy: {accuracy:.4f}")

    return accuracy
    
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.3, random_state=42)
```

## 1. SVM With Train-Test Split (70/30)  

```python
svm = SVC(random_state=42)
svm.fit(X_train, y_train)
svm_tt = evaluate_model(svm, X_test, y_test, "SVM (Train-Test split)")
```

## 2. SVM with 10-fold Cross-Validation

```python
svm_cv = SVC(random_state=42)
cv_scores = cross_val_score(svm_cv, X_scaled, y, cv=10)
svm_cv = cv_scores.mean()
print(f"\nSVM with 10-fold Cross-Validation:")
print(f"CV Accuracy: {svm_cv:.4f}")
```

## 3. SVM with Hyperparameter Tuning

```python
param_grid = {
    'C' : [0.1, 1, 10, 100],
    'gamma': ['scale', 'auto', 0.01, 0.001],
    'kernel': ['rbf', 'linear']
}
grid_search = GridSearchCV(SVC(random_state=42), param_grid, cv=5, scoring='accuracy')
grid_search.fit(X_train, y_train)

print("\nSVM Hyperparameter Tuning Results:")
print(f"Best parameters: {grid_search.best_params_}")
print(f"Best cross-validation score: {grid_search.best_score_:.4f}")

best_svm = grid_search.best_estimator_
svm_tuned_tt = evaluate_model(best_svm, X_test, y_test, "SVM Tuned (Train-Test Split)")
svm_tuned_cv = grid_search.best_score_
```

## 4. SVM with Feature Selection

```python
selector = SelectKBest(f_classif, k=10)
X_train_selected = selector.fit_transform(X_train, y_train)
X_test_selected = selector.transform(X_test)

# Get selected feature names for reference
selected_indices = selector.get_support(indices=True)
selected_features = X.columns[selected_indices]
print("\nSelected Features:")
print(selected_features.tolist())

best_svm_fs = SVC(**grid_search.best_params_, random_state=42)
best_svm_fs.fit(X_train_selected, y_train)
svm_fs_tt = evaluate_model(best_svm_fs, X_test_selected, y_test, "SVM Tuned with Feature Selection")
```

## 5. SVM with PCA

```python
pca = PCA(n_components=10)
X_train_pca = pca.fit_transform(X_train)
X_test_pca = pca.transform(X_test)

print(f"\nPCA Explained Variance: {sum(pca.explained_variance_ratio_):.4f}")

best_svm_pca = SVC(**grid_search.best_params_, random_state=42)
best_svm_pca.fit(X_train_pca, y_train)
svm_pca_tt = evaluate_model(best_svm_pca, X_test_pca, y_test, "SVM Tuned with PCA")
```

## SVM Summary Table

```python
svm_summary = pd.DataFrame({
    'SVM model': [
        ' Train-Test split (70/30)',
        ' 10-fold cross-validation',
        ' 1 and 2 with hyperparameter tuning',
        ' 1 and 2 with hyperparameter tuning and 10 best features',
        ' 1 and 2 with hyperparameter tuning and 10 principal components'
    ],
    'Train-test split': [
        f"{svm_tt:.2%}",
        "N/A",
        f"{svm_tuned_tt:.2%}",
        f"{svm_fs_tt:.2%}",
        f"{svm_pca_tt:.2%}"
    ],
    'Cross-validation': [
        "N/A",
        f"{svm_cv:.2%}",
        f"{svm_tuned_cv:.2%}",
        f"{svm_tuned_cv:.2%}",  
        f"{svm_tuned_cv:.2%}"   
    ]
})

print("\nSVM Models Summary Table")
svm_summary
```
